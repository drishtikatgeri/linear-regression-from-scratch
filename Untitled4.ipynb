# Linear Regression from Scratch (NumPy)

I made fake house data where the price depends on size, bedrooms and age, then taught a model to predict the price using only NumPy. I did it two ways: the normal equation and gradient descent. No scikit-learn for the model itself.

---



```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)

n = 500
sqft = np.random.uniform(500, 3500, n)
bedrooms = np.random.randint(1, 6, n)
age = np.random.uniform(0, 40, n)
noise = np.random.normal(0, 20000, n)

price = 50000 + 150*sqft + 10000*bedrooms - 1000*age + noise
X = np.column_stack([sqft, bedrooms, age])
y = price
```

```python
idx = np.random.permutation(n)
split = int(0.8*n)
tr, te = idx[:split], idx[split:]

mu, sd = X[tr].mean(axis=0), X[tr].std(axis=0)
Xtr = (X[tr] - mu) / sd
Xte = (X[te] - mu) / sd          # scale test using TRAIN stats
Xtr = np.c_[np.ones(len(tr)), Xtr]   # bias column
Xte = np.c_[np.ones(len(te)), Xte]
ytr, yte = y[tr], y[te]
```

```python
A = Xtr.T @ Xtr
print("det:", np.linalg.det(A))   # nonzero means invertible
theta_ne = np.linalg.inv(A) @ Xtr.T @ ytr
```

```python
def gradient_descent(X, y, lr=0.1, epochs=1000):
    m = len(y)
    theta = np.zeros(X.shape[1])
    losses = []
    for _ in range(epochs):
        grad = (2/m) * X.T @ (X @ theta - y)
        theta -= lr * grad
        losses.append(np.mean((X @ theta - y)**2))
    return theta, losses

theta_gd, losses = gradient_descent(Xtr, ytr)
```

```python
def rmse(y, p): return np.sqrt(np.mean((y-p)**2))
def r2(y, p): return 1 - np.sum((y-p)**2) / np.sum((y-y.mean())**2)

for name, th in [("Normal eq", theta_ne), ("Grad descent", theta_gd)]:
    p = Xte @ th
    print(name, "RMSE:", rmse(yte, p), "R2:", r2(yte, p))
```

```python
p = Xte @ theta_ne
plt.scatter(yte, p)
plt.plot([yte.min(), yte.max()], [yte.min(), yte.max()], color='red')
plt.xlabel("Actual price")
plt.ylabel("Predicted price")
plt.title("Predicted vs actual")
plt.show()
```

```python
plt.plot(losses)
plt.xlabel("Epoch (number of tries)")
plt.ylabel("Error")
plt.title("Error going down as the model learns")
plt.show()
```

## Results
- RMSE: ~19,800 and R²: ~0.98 on the test data
- Normal equation and gradient descent gave almost the same answer
- The fake data had noise of about 20,000 built in, so the model can't really do better than that

## What I learned
- Linear regression is just finding the best line (or plane) through the data
- Scaling features matters for gradient descent
- Two completely different methods can reach the same result
